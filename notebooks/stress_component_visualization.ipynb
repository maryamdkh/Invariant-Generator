# Stress-component visualization

A standalone notebook for direct 2D or 3D projections of yield-surface samples. A projection shows every sample after dropping the omitted stress components; it is **not** a fixed-component slice of the yield surface.

Canonical component order: `[s11, s22, s33, s23, s13, s12]`.

```python
from pathlib import Path

from invariant_generator.config import load_config
from invariant_generator.data import canonicalize_stress_features, load_hdf_dataset
from invariant_generator.stress_viz import plot_selected_stress_components

CONFIG_PATH = Path('../configs/default.toml')
config = load_config(CONFIG_PATH)
X_raw = load_hdf_dataset(config.data.data_dir, config.data.dataset_name, dataset_key=config.data.dataset_key)
stress, names = canonicalize_stress_features(X_raw, stress_format=config.data.stress_format)
names, stress.shape
```

```python
# Choose two indices for a 2D projection, e.g. s11 vs s22.
fig = plot_selected_stress_components(
    stress, [0, 1], title='Yield-surface samples: s11 vs s22', max_samples=5000
)
fig
```

```python
# Choose three indices for a 3D projection, e.g. normal stresses.
fig = plot_selected_stress_components(
    stress, [0, 1, 2], title='Yield-surface samples: normal-stress projection', max_samples=5000
)
fig
```

Try `[0, 1, 5]` for a plane-stress-like normal/shear view, or `[3, 4, 5]` for the three shear components. For a physical deviatoric-plane view rather than a coordinate projection, use the existing `scripts/plot_stress_surface.py` diagnostics.
